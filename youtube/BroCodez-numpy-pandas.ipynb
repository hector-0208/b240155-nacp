```python
import pandas as pd

data = [100, 120.2, 104, 200, 232]
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])

print(series)
print(series.loc['c'])
series.loc['c'] = 64
print(series.loc['c'])
print(series.iloc[1])
print(series >= 200)
print(series [series >= 200])
```

```python
calories = {"day 1":1750, "day 2":2100, "day 3":1700}
series = pd.Series(calories)
# series.loc["day 3"] += 500
print(series.loc["day 3"])
print(series[series >= 2000])
print(series[series <= 2000])
```

```python
data = {"Name":["Adam", "Eve", "Lucifer"],
        "Age":[30, 31, 35]}
df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3"])

# print(df.loc["Employee 1"])
# print(df.iloc[1])
# print(df)
df["Job"] = ["Cook", "N/A", "Cashier"]
# print(df)
# print()
new_rows = pd.DataFrame([{"Name":"Dan", "Age":27, "Job":"Engineer"},
                         {"Name":"Bob", "Age":29, "Job":"Waiter"}],
                       index=["Employee 4", "Employee 5"])
df = pd.concat([df, new_rows])
print(df[["Name", "Job"]])
print()
print(df.loc[df["Age"] >= 30, ["Name", "Job"]])
print()
# print(df["Age"].describe())
print(df.loc[df["Age"] == df["Age"].min()])
```
